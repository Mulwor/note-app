```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
from ingest import load_faq_data

documents = load_faq_data()

documents[10]

'''
{
  'id': '310a39c0f9',
  'course': 'ai-dev-tools-zoomcamp',
  'section': 'General Course-Related Questions',
  'question': 'Can I take the course on a locked-down work laptop without admin rights?',
   'answer': 'Yes. Use GitHub Codespaces - you only need a modern browser ..., 
'''
```

```python
# Так как документы состоят из словаря, нам необходимо
# преобразовать его в текст
texts = []

for doc in documents:
  # Берем и объединяем вопрос и ответ
  text = doc["question"] + " " + doc["answer"]
  texts.append(text)
```

```python
texts[10]

'''
  Can I take the course on a locked-down work laptop without admin rights? 
  Yes. Use GitHub Codespaces - you only need a mode ...'
'''
```

```python
len(texts)   # 1400
```

```python
# Теперь нам необходимо преобразовать в векторы
# Однако мы загрузим не все, так как они тяжеловесны, а
# разделим эмбадинги на пакеты

from tqdm.auto import tqdm

batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
  batch = texts[i:i + batch_size]
  batch_vectors = model.encode(batch)
  vectors.extend(batch_vectors)

len(vectors)
```
