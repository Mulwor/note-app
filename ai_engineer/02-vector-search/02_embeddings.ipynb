```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
question_1 = "Can I still join the course after the start date?"
question_2 = "How to install Docker on Windows?"
```

```python
# Кодируем элементы
v1 = model.encode(question_1)
v2 = model.encode(question_2)

# array([ 2.13904064e-02, -7.39799812e-02,  1.42067927e-03,  2.13816669e-02 ...]
```

```python
# Показывает размерность вектора => (384, )
v1.shape   
v2.shape
```

```python
d  = "You don't need to register. You're accepted. You can also just start learning and submitting homework without registering."
dv = model.encode(d)
```

```python
v1.dot(v2)    

# np.float32(-0.14271772) - результат скалярного произведения (dot product) двух векторов.
# Результат -0.14 означает, что модель считает эти тексты семантически далекими друг от друга.
```

```python
# Тестируем векторый поиск

v1.dot(dv)
# np.float32(0.323324), где 0.3 сходства есть на одном вектор

v2.dot(dv)
# np.float32(0.019730467), где 0.01 означат что сходство совсем разные. 
# Другими словами у них нет ничего общего
```
