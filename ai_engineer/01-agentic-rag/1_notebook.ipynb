```python
# Build prompt -  — это мост, который превращает сырые данные (результаты 
# поиска) в понятный для LLM запрос. Без неё LLM либо не увидит ваши документы, 
# либо не поймёт, как их использовать.

# Вот 4 ключевые причины, зачем их создавать:

# 1. Структурирование информации (результат поиск - это список словарей 
# с полями (раздел, вопрос, ответ). LLM работает с текстом, а не 
# со словарями. Функция build_context превращает этот список в одну 
# читаемую строку с чёткими блоками (раздел → Q → A). Так LLM легче
# усваивает данные.

# 2. Добавление инструкций В инструкциях (INSTRUCTIONS) вы говорите LLM: 
# «Отвечай только по контексту, если не знаешь — скажи "I don't know"». 
# Без этих инструкций модель может начать фантазировать (галлюцинировать) 
# или отвечать из своей внутренней памяти, игнорируя ваши документы.

# 3. Разделение фиксированной и изменяемой части 
# - Инструкции (системная подсказка) — постоянны для всех вопросов. 
# - Подсказка пользователя — каждый раз новая (вопрос + контекст). Разде-
# ление позволяет не дублировать инструкции в каждом запросе и легко 
# собирать конечный промпт.

# 4. Контроль формата ответа - В шаблоне USER_PROMPT_TEMPLATE вы 
# явно указываете: сначала вопрос, потом контекст. Это помогает 
# LLM понять, что именно искать, и выдавать ответ в ожидаемом виде.

```

```python
# Программа LLM не видит наши документы, если мы их не передадим. Поэтому 
# нам нужно создать запрос, который будет включать вопрос пользователя и 
# результаты поиска. При создании систем искусственного интеллекта мы 
# обычно разделяем задачу на две части:

# - Инструкции (также называемые системной подсказкой): они указывают LLM, 
# как себя вести. Они никогда не меняются, поэтому одинаковы для каждого 
# запроса.
# - Подсказка пользователю: она меняется с каждым запросом. Она содержит 
# сам вопрос и полученный контекст.

# Мы разделили их, потому что инструкции фиксированы, а подсказки для 
# пользователя — нет. Разделение позволяет легко повторно использовать 
# фиксированную часть и легко создавать новую, изменяемую.
```

```python
# В инструкциях указывается роль магистра права и порядок ответа:

INSTRUCTION = '''
Your task is to answer questions from the course participants
based on the provided context.

Use the context to find relevant information and provide accurate
answers. If the answer is not found in the context,
respond with "I don't know."
'''

# Именно это позволяет подтвердить ответ данными и уменьшить 
# количество галлюцинаци
```

```python
# Создание промпта
USER_PROMPT_TEMPLATE = f'''
Question:
{question}

Context:
{context}
'''
```

```python
# Это context отформатированная строка со всеми результатами поиска:
# Каждый документ становится блоком, содержащим раздел, вопрос и ответ. 
# Такой формат облегчает чтение для LLM. Мы преобразовали список 
# словарей в одну строку. Это небольшой этап предварительной обработки перед отправкой данных в LLM.
def build_context(search_results):
  lines = []

  for doc in search_results:
      lines.append(doc["section"])
      lines.append("Q: " + doc["question"])
      lines.append("A: " + doc["answer"])
      lines.append("")

  return "\n".join(lines).strip()
```

```python
# Теперь объединим prompt с контекстом в подсказку для пользователя:
def build_prompt(question, search_results):
  context = build_context(search_results)
  prompt = USER_PROMPT_TEMPLATE.format(
    question = question,
    context = context
  )
  return prompt.strip()
```

```python
prompt = build_prompt(question, search_results)

print(prompt)
```

```python
# Build prompt -  — это мост, который превращает сырые данные (результаты 
# поиска) в понятный для LLM запрос. Без неё LLM либо не увидит ваши документы, 
# либо не поймёт, как их использовать.

# Вот 4 ключевые причины, зачем их создавать:

# 1. Структурирование информации (результат поиск - это список словарей 
# с полями (раздел, вопрос, ответ). LLM работает с текстом, а не со словарями
# Функция build_context превращает этот список в одну читаемую строку с
# чёткими блоками (раздел → Q → A). Так LLM легче усваивает данные.

# 2. Добавление инструкций В инструкциях (INSTRUCTIONS) вы говорите LLM: 
# «Отвечай только по контексту, если не знаешь — скажи "I don't know"». 
# Без этих инструкций модель может начать фантазировать (галлюцинировать) 
# или отвечать из своей внутренней памяти, игнорируя ваши документы.

# 3. Разделение фиксированной и изменяемой части 
# - Инструкции (системная подсказка) — постоянны для всех вопросов. 
# - Подсказка пользователя — каждый раз новая (вопрос + контекст). Разделение
# позволяет не дублировать инструкции в каждом запросе и легко собирать
# конечный промпт.

# 4. Контроль формата ответа - В шаблоне USER_PROMPT_TEMPLATE вы 
# явно указываете: сначала вопрос, потом контекст. Это помогает 
# LLM понять, что именно искать, и выдавать ответ в ожидаемом виде.

# Пример из вашего файла: Если бы вы просто отправили в LLM голый 
# вопрос «Могу ли я присоединиться к курсу?», она бы ответила из 
# своих общих знаний. А через подсказку вы вставляете реальные 
# FAQ курса, и LLM даёт точный ответ: «Да, но если хотите сертификат,
# сдайте проект до дедлайна».

# Итог: build_prompt — это не просто форматирование, это инструмент 
# управления поведением LLM. Без него ваша RAG-система не будет 
# работать надёжно.
```

```python
# В начале чат гпт под капотом определяет системые требования, то
# есть откуда человек, из какой локации и прочее, затем уже когда 
# вы написали сообщение - он собрал всю необходимо информацию и выда-
# ет вам. Условно как погода? - Если вы в берлине находитесь, то скажет 
# облачно

def llm(instructions, user_prompt, model='gpt-5.4-mini'):
    message_history = [
        {'role': 'developer', 'content': instructions},
        {'role': 'user', 'content': user_prompt}
    ]

    response = openai_client.responses.create(
        model = model,
        input = message_history
    )

    return response.output_text
```

```python
def rag(query, model="gpt-5.4-mini"):
    search_results = search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(INSTRUCTION, prompt, model=model)
    return answer
```
